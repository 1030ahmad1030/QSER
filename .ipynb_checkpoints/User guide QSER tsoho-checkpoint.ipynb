# 2D streamline figure

## Divergence engine

```python
import numpy as np
import matplotlib.pyplot as plt
from QSER.Mesh import Structured1D
import os

fig_path = "/home/ahmad/Yandex.Disk/Research/Q theory/QSER solver 2026/QSER user guide book/figures"
os.makedirs(fig_path, exist_ok=True)

print("=" * 70)
print("GENERATING DIVERGENCE FIGURE")
print("=" * 70)

# Create mesh
mesh = Structured1D(nx=100, L=10.0)
x = mesh.get_cell_centers()
faces = mesh.get_face_centers()

# Create flux field at faces
flux = faces**2

# Compute divergence
div = mesh.compute_divergence(flux)

# Analytical divergence of x^2 is 2x
analytical = 2 * x

# Compute error
interior = slice(2, -2)
error = np.max(np.abs(div[interior] - analytical[interior]))
print(f"Divergence error: {error:.6e}")

# Plot
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Flux
axes[0].plot(faces, flux, 'b-', linewidth=2)
axes[0].set_xlabel('x')
axes[0].set_ylabel('F(x)')
axes[0].set_title('Flux Field: F(x) = x²')
axes[0].grid(True, alpha=0.3)

# Divergence comparison
axes[1].plot(x, analytical, 'k-', linewidth=2, label='Analytical: 2x')
axes[1].plot(x, div, 'r--', linewidth=2, label='Numerical')
axes[1].set_xlabel('x')
axes[1].set_ylabel('dF/dx')
axes[1].set_title('Divergence Comparison')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Error
axes[2].plot(x, np.abs(div - analytical), 'r-', linewidth=2)
axes[2].set_xlabel('x')
axes[2].set_ylabel('Error')
axes[2].set_title('Divergence Error')
axes[2].grid(True, alpha=0.3)
axes[2].set_yscale('log')

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_divergence.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"\nFigure saved: {os.path.join(fig_path, '1d_divergence.png')}")
print("=" * 70)
print("✅ DIVERGENCE FIGURE GENERATED")
print("=" * 70)
```

# Backend codes

```python
import numpy as np
import matplotlib.pyplot as plt
from QSER.Mesh import Structured1D
import time
import os

fig_path = "/home/ahmad/Yandex.Disk/Research/Q theory/QSER solver 2026/QSER user guide book/figures"
os.makedirs(fig_path, exist_ok=True)

print("=" * 70)
print("GENERATING BACKEND COMPARISON FIGURE")
print("=" * 70)

# Test configurations
sizes = [100, 500, 1000, 2000, 5000]
backends = ['numpy', 'torch', 'jax']
colors = {'numpy': 'blue', 'torch': 'orange', 'jax': 'green'}
markers = {'numpy': 'o', 'torch': 's', 'jax': '^'}

results = {backend: [] for backend in backends}
errors = {backend: [] for backend in backends}

for N in sizes:
    print(f"\nN = {N}")
    for backend in backends:
        try:
            # Create mesh
            start_time = time.time()
            mesh = Structured1D(nx=N, L=10.0, backend=backend)
            x = mesh.get_cell_centers()
            field = np.sin(x)
            
            # Compute gradient
            grad = mesh.compute_gradient(field, method='5point')
            elapsed = time.time() - start_time
            
            # Compute error
            analytical = np.cos(x)
            interior = slice(2, -2)
            error = np.max(np.abs(grad[interior] - analytical[interior]))
            
            results[backend].append(elapsed)
            errors[backend].append(error)
            print(f"  {backend:8s}: time = {elapsed:.4f}s, error = {error:.6e}")
        except Exception as e:
            print(f"  {backend:8s}: FAILED - {str(e)}")
            results[backend].append(np.nan)
            errors[backend].append(np.nan)

# Plot speed comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Speed comparison
ax1 = axes[0]
for backend in backends:
    ax1.plot(sizes, results[backend], color=colors[backend], marker=markers[backend], 
             linewidth=2, markersize=8, label=backend.capitalize())
ax1.set_xlabel('Number of Cells', fontsize=12)
ax1.set_ylabel('Computation Time (s)', fontsize=12)
ax1.set_title('Gradient Computation Speed', fontsize=14)
ax1.legend(fontsize=12)
ax1.grid(True, alpha=0.3)

# Error comparison
ax2 = axes[1]
for backend in backends:
    ax2.plot(sizes, errors[backend], color=colors[backend], marker=markers[backend], 
             linewidth=2, markersize=8, label=backend.capitalize())
ax2.set_xlabel('Number of Cells', fontsize=12)
ax2.set_ylabel('Max Error', fontsize=12)
ax2.set_title('Gradient Accuracy', fontsize=14)
ax2.legend(fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.set_yscale('log')

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_backend_comparison.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"\nFigure saved: {os.path.join(fig_path, '1d_backend_comparison.png')}")
print("=" * 70)
print("✅ BACKEND COMPARISON FIGURE GENERATED")
print("=" * 70)
```

## 1D codes section 4.1

```python
import numpy as np
import matplotlib.pyplot as plt
from QSER.Mesh import Structured1D
from QSER.Operators import Gradient, Laplacian, Divergence
from QSER.Mesh.interpolation import Interpolation
from QSER.Mesh.boundary import Boundary
from QSER.Mesh.quality import MeshQuality
import os
import time

# Set figure path
fig_path = "/home/ahmad/Yandex.Disk/Research/Q theory/QSER solver 2026/QSER user guide book/figures"
os.makedirs(fig_path, exist_ok=True)

print("=" * 70)
print("GENERATING FIGURES FOR CHAPTER 4")
print("=" * 70)

# ============================================================
# FIGURE 1: QSER Decomposition (Section 4.1)
# ============================================================
print("\n1. QSER Decomposition")
x = np.linspace(0, 10, 100)
S = np.sin(x) * np.exp(-0.1 * x)
E = 0.3 * x * np.exp(-0.2 * x)
R = S - E

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x, S, 'b-', linewidth=3, label='Source Field $S$ (Ghost)')
ax.plot(x, E, 'r-', linewidth=3, label='Environment Field $E$ (Memory)')
ax.plot(x, R, 'g--', linewidth=3, label='Response $R = S - E$')
ax.set_xlabel('Position $x$', fontsize=14)
ax.set_ylabel('Field Value', fontsize=14)
ax.set_title('QSER Decomposition: $R = S - E$', fontsize=16)
ax.legend(fontsize=12)
ax.grid(True, alpha=0.3)
ax.set_xlim(0, 10)
plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_qser_decomposition.png'), dpi=300, bbox_inches='tight')
plt.close()
print(f"  Saved: {os.path.join(fig_path, '1d_qser_decomposition.png')}")

# ============================================================
# FIGURE 2: Architecture Diagram (Section 4.2)
# ============================================================
print("\n2. Architecture Diagram")
fig, ax = plt.subplots(figsize=(14, 10))
ax.set_xlim(0, 14)
ax.set_ylim(0, 10)
ax.axis('off')

ax.text(7, 9.5, 'QSER 1D Mesh Architecture', fontsize=20, fontweight='bold', ha='center')

# User box
ax.text(7, 8.5, 'User Code', fontsize=14, fontweight='bold', ha='center',
        bbox=dict(boxstyle="round,pad=0.5", facecolor='lightyellow', edgecolor='orange', linewidth=2))

# Arrow
ax.annotate('', xy=(7, 7.8), xytext=(7, 8.0), arrowprops=dict(arrowstyle='->', lw=2))

# Mesh box
ax.text(7, 7.2, 'Structured1D', fontsize=14, fontweight='bold', ha='center',
        bbox=dict(boxstyle="round,pad=0.5", facecolor='lightblue', edgecolor='blue', linewidth=2))

# Engines
engines = [('Gradient', 2, 6.0), ('Laplacian', 4.5, 6.0), ('Divergence', 7, 6.0),
           ('Interpolation', 9.5, 6.0), ('Boundary', 12, 6.0)]
for name, ex, ey in engines:
    ax.text(ex, ey, name, fontsize=12, fontweight='bold', ha='center',
            bbox=dict(boxstyle="round,pad=0.4", facecolor='lightgreen', edgecolor='green', linewidth=2))

# Backends
backends = [('NumPy', 2, 4.5), ('PyTorch', 5, 4.5), ('JAX', 8, 4.5), ('OpenFOAM', 11, 4.5)]
for name, bx, by in backends:
    ax.text(bx, by, name, fontsize=12, fontweight='bold', ha='center',
            bbox=dict(boxstyle="round,pad=0.4", facecolor='lightcoral', edgecolor='red', linewidth=2))

# Hardware
ax.text(7, 3.0, 'Hardware (CPU / GPU / TPU)', fontsize=12, fontweight='bold', ha='center',
        bbox=dict(boxstyle="round,pad=0.5", facecolor='lightgray', edgecolor='gray', linewidth=2))

# Arrows
for ex, ey in [(2, 6.0), (4.5, 6.0), (7, 6.0), (9.5, 6.0), (12, 6.0)]:
    ax.annotate('', xy=(ex, 6.4), xytext=(ex, 6.8), arrowprops=dict(arrowstyle='->', lw=2))
for bx, by in [(2, 4.5), (5, 4.5), (8, 4.5), (11, 4.5)]:
    ax.annotate('', xy=(bx, 4.9), xytext=(bx, 5.6), arrowprops=dict(arrowstyle='->', lw=2))
for bx, by in [(2, 4.5), (5, 4.5), (8, 4.5), (11, 4.5)]:
    ax.annotate('', xy=(7, 3.4), xytext=(bx, 4.1), arrowprops=dict(arrowstyle='->', lw=1, linestyle='dashed'))

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_architecture.png'), dpi=300, bbox_inches='tight')
plt.close()
print(f"  Saved: {os.path.join(fig_path, '1d_architecture.png')}")

# ============================================================
# FIGURE 3: Mesh Plots (Section 4.4)
# ============================================================
print("\n3. Mesh Plots")
mesh_uniform = Structured1D(nx=20, L=10.0)
mesh_nonuniform = Structured1D(nx=20, L=10.0, uniform=False)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, mesh, title in zip(axes, [mesh_uniform, mesh_nonuniform],
                           ['Uniform Mesh', 'Non-Uniform Mesh']):
    x_np = mesh.get_face_centers()
    centers_np = mesh.get_cell_centers()
    ax.plot(x_np, np.zeros_like(x_np), 'ko-', markersize=6, label='Nodes')
    ax.plot(centers_np, np.zeros_like(centers_np), 'r^', markersize=8, label='Cell Centers')
    ax.set_xlabel('x')
    ax.set_ylabel('')
    ax.set_title(title)
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_mesh.png'), dpi=300, bbox_inches='tight')
plt.close()
print(f"  Saved: {os.path.join(fig_path, '1d_mesh.png')}")

# ============================================================
# FIGURE 4: Gradient Comparison (Section 4.6)
# ============================================================
print("\n4. Gradient Comparison")
mesh = Structured1D(nx=100, L=10.0)
x = mesh.get_cell_centers()
field = np.sin(x)
analytical = np.cos(x)

grad_5p = mesh.compute_gradient(field, method='5point')
grad_ls = mesh.compute_gradient(field, method='least_squares')

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].plot(x, field, 'b-', linewidth=2)
axes[0].set_xlabel('x')
axes[0].set_ylabel('f(x)')
axes[0].set_title('Field: sin(x)')
axes[0].grid(True, alpha=0.3)

axes[1].plot(x, analytical, 'k-', linewidth=2, label='Analytical')
axes[1].plot(x, grad_5p, 'r--', linewidth=2, label='5-Point')
axes[1].plot(x, grad_ls, 'g--', linewidth=2, label='Least-Squares')
axes[1].set_xlabel('x')
axes[1].set_ylabel('df/dx')
axes[1].set_title('Gradient Comparison')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

axes[2].plot(x, np.abs(grad_5p - analytical), 'r-', label='5-Point Error')
axes[2].plot(x, np.abs(grad_ls - analytical), 'g-', label='LS Error')
axes[2].set_xlabel('x')
axes[2].set_ylabel('Error')
axes[2].set_title('Gradient Errors')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_gradient_comparison.png'), dpi=300, bbox_inches='tight')
plt.close()
print(f"  Saved: {os.path.join(fig_path, '1d_gradient_comparison.png')}")

# ============================================================
# FIGURE 5: Laplacian Comparison (Section 4.7)
# ============================================================
print("\n5. Laplacian Comparison")
lap_5p = mesh.compute_laplacian(field)
analytical_lap = -np.sin(x)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].plot(x, field, 'b-', linewidth=2)
axes[0].set_xlabel('x')
axes[0].set_ylabel('f(x)')
axes[0].set_title('Field: sin(x)')
axes[0].grid(True, alpha=0.3)

axes[1].plot(x, analytical_lap, 'k-', linewidth=2, label='Analytical')
axes[1].plot(x, lap_5p, 'r--', linewidth=2, label='Numerical')
axes[1].set_xlabel('x')
axes[1].set_ylabel('d²f/dx²')
axes[1].set_title('Laplacian Comparison')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

axes[2].plot(x, np.abs(lap_5p - analytical_lap), 'r-')
axes[2].set_xlabel('x')
axes[2].set_ylabel('Error')
axes[2].set_title('Laplacian Error')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_laplacian_comparison.png'), dpi=300, bbox_inches='tight')
plt.close()
print(f"  Saved: {os.path.join(fig_path, '1d_laplacian_comparison.png')}")

# ============================================================
# FIGURE 6: Interpolation Comparison (Section 4.9)
# ============================================================
print("\n6. Interpolation Comparison")
faces = mesh.get_face_centers()
methods = ['linear', 'nearest', 'cubic', 'pchip', 'savgol']
colors = ['r', 'orange', 'g', 'm', 'brown']

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Interpolation comparison
ax = axes[0]
ax.plot(x, field, 'b-', linewidth=2, label='Original (cell centers)')
for method, color in zip(methods, colors):
    interp = mesh.interpolate_to_faces(field, method=method)
    ax.plot(faces, interp, '--', color=color, linewidth=1.5, label=method.capitalize())
ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.set_title('Interpolation Methods Comparison')
ax.legend(loc='upper right', fontsize=8)
ax.grid(True, alpha=0.3)

# Errors
ax2 = axes[1]
for method, color in zip(methods, colors):
    interp = mesh.interpolate_to_faces(field, method=method)
    error = np.abs(interp - np.sin(faces))
    ax2.plot(faces, error, '-', color=color, linewidth=1.5, label=method.capitalize())
ax2.set_xlabel('x')
ax2.set_ylabel('Error')
ax2.set_title('Interpolation Errors')
ax2.legend(loc='upper left', fontsize=8)
ax2.grid(True, alpha=0.3)

# Max errors
ax3 = axes[2]
max_errors = []
for method in methods:
    interp = mesh.interpolate_to_faces(field, method=method)
    max_errors.append(np.max(np.abs(interp - np.sin(faces))))
ax3.bar(methods, max_errors, color=colors)
ax3.set_ylabel('Max Error')
ax3.set_title('Max Interpolation Error')
ax3.set_yscale('log')
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_interpolation_comparison.png'), dpi=300, bbox_inches='tight')
plt.close()
print(f"  Saved: {os.path.join(fig_path, '1d_interpolation_comparison.png')}")

# ============================================================
# FIGURE 7: Boundary Conditions (Section 4.10)
# ============================================================
print("\n7. Boundary Conditions")
boundary = Boundary(mesh)
field_bc = mesh.apply_bc(field, 'dirichlet', [0.0, 1.0])

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Original
axes[0].plot(x, field, 'b-', linewidth=2)
axes[0].axvline(x[0], color='red', linestyle='--', alpha=0.5)
axes[0].axvline(x[-1], color='red', linestyle='--', alpha=0.5)
axes[0].set_xlabel('x')
axes[0].set_ylabel('f(x)')
axes[0].set_title('Original Field')
axes[0].grid(True, alpha=0.3)

# Dirichlet
field_dir = mesh.apply_bc(field, 'dirichlet', [0.0, 1.0])
axes[1].plot(x, field_dir, 'r-', linewidth=2)
axes[1].axvline(x[0], color='red', linestyle='--', alpha=0.5)
axes[1].axvline(x[-1], color='red', linestyle='--', alpha=0.5)
axes[1].set_xlabel('x')
axes[1].set_ylabel('f(x)')
axes[1].set_title('Dirichlet BC: f(0)=0, f(L)=1')
axes[1].grid(True, alpha=0.3)

# Neumann
field_neu = mesh.apply_bc(field, 'neumann', [0.0, 0.0])
axes[2].plot(x, field_neu, 'g-', linewidth=2)
axes[2].axvline(x[0], color='red', linestyle='--', alpha=0.5)
axes[2].axvline(x[-1], color='red', linestyle='--', alpha=0.5)
axes[2].set_xlabel('x')
axes[2].set_ylabel('f(x)')
axes[2].set_title('Neumann BC: df/dn=0')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_boundary_conditions.png'), dpi=300, bbox_inches='tight')
plt.close()
print(f"  Saved: {os.path.join(fig_path, '1d_boundary_conditions.png')}")

# ============================================================
# FIGURE 8: Convergence (Section 4.13)
# ============================================================
print("\n8. Convergence")
Ns = [20, 50, 100, 200, 500]
errors_5p = []
errors_ls = []

for N in Ns:
    mesh_N = Structured1D(nx=N, L=10.0)
    x_N = mesh_N.get_cell_centers()
    field_N = np.sin(x_N)
    grad_N = mesh_N.compute_gradient(field_N, method='5point')
    grad_ls_N = mesh_N.compute_gradient(field_N, method='least_squares')
    analytical_N = np.cos(x_N)
    interior = slice(2, -2)
    errors_5p.append(np.max(np.abs(grad_N[interior] - analytical_N[interior])))
    errors_ls.append(np.max(np.abs(grad_ls_N[interior] - analytical_N[interior])))

fig, ax = plt.subplots(figsize=(8, 6))
ax.loglog(Ns, errors_5p, 'ro-', linewidth=2, label='5-Point')
ax.loglog(Ns, errors_ls, 'bs-', linewidth=2, label='Least-Squares')
ax.loglog(Ns, 1 / np.array(Ns)**2, 'k--', alpha=0.5, label='O(N⁻²)')
ax.loglog(Ns, 1 / np.array(Ns)**4, 'k:', alpha=0.5, label='O(N⁻⁴)')
ax.set_xlabel('Number of Cells N')
ax.set_ylabel('Max Error')
ax.set_title('Convergence of Gradient Methods')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(fig_path, '1d_convergence.png'), dpi=300, bbox_inches='tight')
plt.close()
print(f"  Saved: {os.path.join(fig_path, '1d_convergence.png')}")

print("\n" + "=" * 70)
print("✅ ALL FIGURES GENERATED")
print("=" * 70)
print(f"Figures saved to: {fig_path}")
```
