# Structured 1d mesh function

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
from scipy.interpolate import PchipInterpolator, CubicSpline

class Structured1D:
    """
    1D Structured Mesh with multiple interpolation options.
    """

    def __init__(self, nx=100, L=1.0, uniform=True, x=None):
        self.dim = 1
        self.nx = nx
        self.n_cells = nx
        self.n_nodes = nx + 1
        self.L = L

        if x is not None:
            self.x = np.array(x)
            self.nx = len(x) - 1
            self.n_cells = self.nx
            self.n_nodes = len(x)
            self.L = x[-1] - x[0]
        elif uniform:
            self.x = np.linspace(0, L, nx + 1)
        else:
            self.x = np.linspace(0, L, nx + 1)**2 / L

        self.cell_centers = (self.x[:-1] + self.x[1:]) / 2
        self.cell_volumes = self.x[1:] - self.x[:-1]
        self.face_centers = self.x
        self.n_faces = self.n_nodes

    def get_cell_centers(self):
        return self.cell_centers

    def get_cell_volumes(self):
        return self.cell_volumes

    def get_face_centers(self):
        return self.face_centers

    def compute_gradient(self, field):
        grad = np.zeros(self.n_cells)
        for i in range(1, self.n_cells - 1):
            grad[i] = (field[i+1] - field[i-1]) / (self.cell_centers[i+1] - self.cell_centers[i-1])
        grad[0] = (field[1] - field[0]) / self.cell_volumes[0]
        grad[-1] = (field[-1] - field[-2]) / self.cell_volumes[-1]
        return grad

    def compute_laplacian(self, field):
        lap = np.zeros(self.n_cells)
        for i in range(1, self.n_cells - 1):
            lap[i] = (field[i+1] - 2*field[i] + field[i-1]) / self.cell_volumes[i]**2
        lap[0] = (field[1] - 2*field[0] + field[0]) / self.cell_volumes[0]**2
        lap[-1] = (field[-1] - 2*field[-1] + field[-2]) / self.cell_volumes[-1]**2
        return lap

    def interpolate_to_faces(self, field, method='linear'):
        """
        Interpolate cell-centered field to faces.

        Parameters:
            field: (n_cells,) array
            method: str, interpolation method
                'linear'   -> linear interpolation (default)
                'nearest'  -> nearest neighbour
                'cubic'    -> cubic spline (scipy)
                'savgol'   -> Savitzky-Golay filter (scipy)
                'pchip'    -> monotonic cubic (scipy)

        Returns:
            face_field: (n_faces,) array
        """
        face_field = np.zeros(self.n_faces)

        if method == 'linear':
            # Linear interpolation
            for i in range(1, self.n_faces - 1):
                face_field[i] = (field[i-1] + field[i]) / 2
            face_field[0] = field[0]
            face_field[-1] = field[-1]

        elif method == 'nearest':
            # Nearest neighbour
            for i in range(self.n_faces):
                if i == 0:
                    face_field[i] = field[0]
                elif i == self.n_faces - 1:
                    face_field[i] = field[-1]
                else:
                    # Pick the closer cell centre
                    if (self.face_centers[i] - self.cell_centers[i-1]) < (self.cell_centers[i] - self.face_centers[i]):
                        face_field[i] = field[i-1]
                    else:
                        face_field[i] = field[i]

        elif method == 'cubic':
            # Cubic spline (requires scipy)
            from scipy.interpolate import CubicSpline
            cs = CubicSpline(self.cell_centers, field, bc_type='natural')
            face_field = cs(self.face_centers)

        elif method == 'savgol':
            # Savitzky-Golay filter (requires scipy)
            from scipy.signal import savgol_filter
            # Need to ensure we have enough points
            window_length = min(len(field) // 2 * 2 + 1, 11)
            if window_length < 3:
                window_length = 3
            smoothed = savgol_filter(field, window_length, 3)
            # Interpolate smoothed to faces
            for i in range(1, self.n_faces - 1):
                face_field[i] = (smoothed[i-1] + smoothed[i]) / 2
            face_field[0] = smoothed[0]
            face_field[-1] = smoothed[-1]

        elif method == 'pchip':
            # PCHIP (monotonic cubic)
            from scipy.interpolate import PchipInterpolator
            pchip = PchipInterpolator(self.cell_centers, field)
            face_field = pchip(self.face_centers)

        else:
            raise ValueError(f"Unknown interpolation method: {method}")

        return face_field

    def apply_bc(self, field, bc_type, bc_value):
        if bc_type == 'dirichlet':
            field[0] = bc_value[0]
            field[-1] = bc_value[1]
        elif bc_type == 'neumann':
            field[0] = field[1] - bc_value[0] * self.cell_volumes[0]
            field[-1] = field[-2] + bc_value[1] * self.cell_volumes[-1]
        return field

    def plot(self, field=None, title="1D Mesh"):
        plt.figure(figsize=(10, 4))
        plt.plot(self.x, np.zeros_like(self.x), 'ko-', markersize=4, label='Nodes')
        plt.plot(self.cell_centers, np.zeros_like(self.cell_centers), 'r^', markersize=6, label='Cell Centers')
        if field is not None:
            plt.plot(self.cell_centers, field, 'b-', linewidth=2, label='Field')
        plt.xlabel('x (m)')
        plt.ylabel('Value')
        plt.title(title)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()
```

### Testing 1d mesh

```python
# Create mesh
mesh = Structured1D(nx=20, L=10.0)
x = mesh.get_cell_centers()
field = np.exp(-(x - 5.0)**2 / 0.5)

# Test all interpolation methods
methods = ['linear', 'nearest', 'cubic', 'savgol', 'pchip']

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for idx, method in enumerate(methods):
    ax = axes[idx]
    face_field = mesh.interpolate_to_faces(field, method=method)

    ax.plot(mesh.cell_centers, field, 'bo-', label='Cell centers', markersize=6)
    ax.plot(mesh.face_centers, face_field, 'r.-', label=f'{method} interpolation', markersize=8)
    ax.set_title(f'Method: {method}')
    ax.set_xlabel('x')
    ax.set_ylabel('Value')
    ax.legend()
    ax.grid(True, alpha=0.3)

# Remove empty subplot (if any)
if len(methods) < len(axes):
    for i in range(len(methods), len(axes)):
        fig.delaxes(axes[i])

plt.tight_layout()
plt.show()
```
